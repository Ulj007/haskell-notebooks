# Övning 3: Rekursion

En **rekursiv funktion** är en funktion som är definierad i termer av sig själv.

I de följande uppgifterna får du öva på att implementera funktioner som använder sig av rekursion. Om inget annat är angivet ska funktionen som du definierar alltså innehålla ett fall där den anropar sig själv.

## Lärardemo 1: Fakultet

Fakultet är en matematisk funktion som framför allt används inom sannolikhetslära och kombinatorik. Fakulteten betecknas med symbolen $!$.

För ett heltal $n>0$ är

$$n! = n(n-1)(n-2)\cdots 2\cdot1$$

Till exempel:

$$4! = 4\cdot3\cdot2\cdot1 = 24$$

Dessutom är $0!$ definierad som $0! = 1$.

**Uppgift:** Definiera en rekursiv funktion `factorial` (med signatur) som räknar ut fakulteten för ett givet heltal.


````haskell
factorial :: (Integral a) => a -> a
factorial 0 = 1
factorial n
    | n < 0     = error "negative integral"
    | otherwise = n * factorial (n-1)
````

````haskell
factorial 4
factorial 10
````

````haskell
--factorial (-1)
````

### Vad händer vid `factorial 4`?

Varje anrop reducerar problemet genom att anropa `factorial` med ett mindre tal:

$$
\begin{aligned}
\texttt{factorial 4}
&= 4 \cdot \texttt{factorial 3}\\
&= 4 \cdot (3 \cdot \texttt{factorial 2})\\
&= 4 \cdot (3 \cdot (2 \cdot \texttt{factorial 1}))\\
&= 4 \cdot (3 \cdot (2 \cdot (1 \cdot \texttt{factorial 0})))\\
&= 4 \cdot (3 \cdot (2 \cdot (1 \cdot 1)))\\
&= 24
\end{aligned}
$$

**Basfallet** `factorial 0 = 1` gör att rekursionen avslutas.

## 2. Längden på en lista

Definiera en rekursiv funktion `listLength` (med signatur), som räknar ut längden på en godtycklig lista.

Funktionen finns redan definierad i `Prelude` under namnet `length`. Du kan därför testa din funktion genom att se om den ger samma svar som `length`.


````haskell
listLength :: [a] -> Int
listLength []     = 0
listLength (_:xs) = 1 + listLength xs
````

### Vad händer vid `listLength [1,2,3]`?

Funktionen tar bort ett element i taget och anropar sig själv med resten av listan:

```text
listLength [1,2,3]
= 1 + listLength [2,3]
= 1 + (1 + listLength [3])
= 1 + (1 + (1 + listLength []))
= 1 + (1 + (1 + 0))
= 3

````haskell
listLength [1..10]
listLength ['a'..'c']
listLength [1,2,3] == length [1,2,3]
````

## 3. Inga jämna tal

Definiera funktionen `noEvens` (med signatur), som rekursivt testar att en heltalslista inte innehåller något jämnt tal.


````haskell
noEvens :: (Integral a) => [a] -> Bool
noEvens [] = True
noEvens (x:xs)
    | even x    = False
    | otherwise = noEvens xs
````

````haskell
noEvens [1,2,3]
noEvens [1,3..9]
````

## 4. Antal siffror

Definiera en rekursiv funktion `digits` (med signatur), som returnerar antalet siffror i ett givet heltal.

**Obs!** Funktionen ska fungera även för negativa heltal.


````haskell
digits :: Int -> Int
digits x
    | x < 0     = digits (-x)
    | x < 10    = 1
    | otherwise = 1 + digits (x `div` 10)
````

````haskell
digits 123
digits (-10000)
````

## 5. Största summan av tre tal i följd

Definiera den rekursiva funktionen `max3Sequence` (med signatur), som returnerar den största möjliga summan av tre direkt efter varandra följande tal i en lista.


````haskell
max3Sequence :: (Num a, Ord a) => [a] -> a
max3Sequence []     = 0
max3Sequence all@(x:xs)
    | now >= later  = now
    | otherwise     = later
    where now   = sum (take 3 all)
          later = max3Sequence xs
````

````haskell
max3Sequence [1,2,3,4,3,2,1]
max3Sequence [-10,10,-10,1,1,1]
````

## 6. Generera en oändlig lista

Definiera en rekursiv funktion `generateInfiniteList` (med signatur), som tar två heltal som parametrar: ett startvärde `n` samt ett värde `m` som bestämmer stegavståndet.

`generateInfiniteList n m` ska ge den obegränsade sekvensen

```text
[n,(n+m)..]
```

**Obs!** Testa den obegränsade listan med `take`, exempelvis `take 6 (generateInfiniteList 10 3)`.


````haskell
generateInfiniteList :: Integer -> Integer -> [Integer]
generateInfiniteList n m = n : generateInfiniteList (n+m) m
````

````haskell
take 6 (generateInfiniteList 10 3)
````

## 7. Kumulativ lista

Definiera `cumulativeList` (med signatur), som är en oändlig heltalslista som börjar med talet 1 och där varje nästkommande element är summan av alla tidigare element.


````haskell
cumulativeList :: [Integer]
cumulativeList = 1 : [sum (take n cumulativeList) | n <- [1..]]
````

````haskell
take 10 cumulativeList
````

## 8. Quicksort

Definiera en rekursiv funktion `quicksort` (med signatur), som tar en ordningsbar lista som parameter och returnerar listan sorterad i stigande ordning.


````haskell
quicksort :: (Ord a) => [a] -> [a]
quicksort []     = []
quicksort (x:xs) =
    let smallerSorted = quicksort [a | a <- xs, a <= x]
        biggerSorted  = quicksort [a | a <- xs, a > x]
    in  smallerSorted ++ [x] ++ biggerSorted
````

### Vad händer vid `quicksort [3,1,4,2]`?

Det första elementet, `3`, används som **pivot**. Resten av listan delas upp i tal som är mindre än eller lika med pivot och tal som är större:

```text
                 [3,1,4,2]
                 pivot = 3
                  /     \
               [1,2]    [4]

               

                 [3,1,4,2]
                  /     \
               [1,2]    [4]
                / \
              []  [2]


quicksort [1,2]
= [] ++ [1] ++ [2]
= [1,2]


quicksort [3,1,4,2]
= [1,2] ++ [3] ++ [4]
= [1,2,3,4]

````haskell
quicksort [10,2,5,3,1,6,7,4,2,3,4,8,9]
quicksort "the quick brown fox jumps over the lazy dog"
````
